```python
import os
from datetime import datetime, timedelta
import papermill as pm
import pandas as pd
```

```python
# action(1/1): : 设置时间范围，不应该超过1d中设置的范围
start_date = datetime(2026, 5, 27)
end_date = datetime(2026, 7, 3)
# 运行全部后，查看MMDD-MMDD.xlsx
```

---

```python
input_nb = "1c_calculate_chain.ipynb"
tmp_csv = "tmp_metrics_temp.csv"  # 放项目本地，跨Windows/Mac/Linux通用
# =================================================

# 生成MMDD列表
mmdd_list = []
cur = start_date
while cur <= end_date:
    mmdd_list.append(cur.strftime("%m%d"))
    cur += timedelta(days=1)

total_df = pd.DataFrame()
start_mmdd = mmdd_list[0]
end_mmdd = mmdd_list[-1]

print(f"开始循环计算，共{len(mmdd_list)}个交易日，输出文件：{start_mmdd}-{end_mmdd}.xlsx")

for mmdd in mmdd_list:
    print(f"\n正在处理 MMDD = {mmdd}")
    temp_nb = f"temp_run_{mmdd}.ipynb"

    try:
        # 直接执行原始1c，不修改源码
        pm.execute_notebook(
            input_path=input_nb,
            output_path=temp_nb,
            parameters={"MMDD": mmdd},
            log_output=False
        )
        print(f"✅ 计算Notebook执行完成")

        # 读取1c输出的本地临时csv
        if os.path.exists(tmp_csv):
            daily_df = pd.read_csv(tmp_csv)
            daily_df["MMDD"] = mmdd
            total_df = pd.concat([total_df, daily_df], ignore_index=True)
            print(f"📊 读取当日指标成功，列数：{daily_df.shape[1]}")
            os.remove(tmp_csv)  # 清理当日临时csv
        else:
            print(f"⚠️ 未找到指标文件 {tmp_csv}")

    except Exception as err:
        print(f"❌ {mmdd} 执行异常：{str(err)}")
    finally:
        # 强制删除临时ipynb
        if os.path.exists(temp_nb):
            os.remove(temp_nb)

# 导出汇总Excel
out_file = f"{start_mmdd}-{end_mmdd}.xlsx"
if not total_df.empty:
    # MMDD放第一列，结构平整无折叠
    cols = ["MMDD"] + [c for c in total_df.columns if c != "MMDD"]
    total_df = total_df[cols]
    total_df.to_excel(out_file, index=False)
    print(f"\n🎉 全部完成！汇总表已导出：{out_file}")
    print(f"总行数：{total_df.shape[0]}，总指标列：{total_df.shape[1]-1}")
else:
    print("\n❌ 无任何有效指标数据，未生成Excel")
```
