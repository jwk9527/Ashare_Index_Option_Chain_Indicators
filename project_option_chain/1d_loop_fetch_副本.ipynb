```python
import os
from datetime import datetime, timedelta
import papermill as pm
```

```python
# action(1/1): 设置时间范围，最多两个合约月
start_date = datetime(2026, 7, 22)
end_date = datetime(2026, 7, 30)
# 运行全部后，进入1e_loop_calc_export.ipynb
```

---

```python
# 生成日期列表
date_list = []
current_date = start_date
while current_date <= end_date:
    date_list.append(current_date.strftime("%Y-%m-%d"))
    current_date += timedelta(days=1)

# 创建最终存放 HTML 报告的文件夹
output_dir = "./html_reports"
os.makedirs(output_dir, exist_ok=True)

print(f"🚀 开始跑批任务，准备生成 {len(date_list)} 份 HTML 报告...")

for text_date in date_list:
    print(f"\n[处理中] 日期: {text_date} ...")

    # 定义临时生成的 ipynb 和最终的 html 路径
    temp_notebook = f"temp_{text_date}.ipynb"
    final_html = f"{output_dir}/report_{text_date}.html"

    try:
        # 步骤 A: 传入日期参数，运行计算程序，生成一个临时 notebook
        pm.execute_notebook(
            input_path="1b_fetch_chain_data.ipynb",
            output_path=temp_notebook,
            parameters=dict(date=text_date),
            log_output=False,  # 隐藏冗长的运行日志
        )

        # 步骤 B: 使用 nbconvert 将临时 notebook 转换为 HTML，并隐藏代码块 (--no-input)
        # os.system 可以直接在 Python 中执行终端命令
        convert_cmd = (
            f"jupyter nbconvert --to html "
            f"--no-input "
            f"--output-dir='{output_dir}' "
            f"--output='report_{text_date}.html' "
            f"{temp_notebook}"
        )
        os.system(convert_cmd)
        print(f" └─ 📄 HTML 报告生成成功: {final_html}")

    except Exception as e:
        print(f" └─ ❌ 运行出错: {e}")

    finally:
        # 步骤 C: 无论成功还是报错，都把临时的 .ipynb 文件删掉，保持彻底干净
        if os.path.exists(temp_notebook):
            os.remove(temp_notebook)

print("\n🎉 所有日期处理完毕！请在 './html_reports' 文件夹中查看您的报告。")
```
